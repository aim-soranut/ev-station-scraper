```python
from pugev import get_pathum_wan_polygon, filter_stations_in_polygon
import json
import pyperclip

with open("pugev_stations.txt", "r", encoding="utf-8") as f:
    all_stations = json.load(f)
```

```python
pathum_wan_polygon = get_pathum_wan_polygon()
pathum_wan_stations = filter_stations_in_polygon(all_stations, pathum_wan_polygon)

print(f"Stations in Pathum Wan: {len(pathum_wan_stations)}")
```

```python
with open("pathum_wan_stations.txt", "r", encoding="utf-8") as f:
    pathum_wan_stations = json.load(f)
```

```python
for station in pathum_wan_stations:
    print(station["name"])
```

```python
import time

for station in pathum_wan_stations:
    print(station["name"])
    print(station["address"])
    time.sleep(0.2)
    id = station.get("id")
    lat = station.get("latitude")
    long = station.get("longitude")
    if not lat or not long:
        print("No full coordinates")
    else:
        coordinates = f"{lat},{long}"
        pyperclip.copy(coordinates)
    input("Press enter to copy lat and long for excel")
    coordinates = f"{lat}\t{long}\t{id}"
    pyperclip.copy(coordinates)
    input("Press enter to continue")
```
